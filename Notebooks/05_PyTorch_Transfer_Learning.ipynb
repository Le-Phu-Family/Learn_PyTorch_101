> Why use Transfer Learning
- Can leverage an existing model (usually a neural network architecture) proven to work on problems similar to our own.
- Can leverage a working model which has already learned patterns on similar data to our own. This often results in achieving great results with less custom data.

- Transfer learning often allows you to get good results with a relatively small amount of custom data.
- Knowing the power of transfer learning, it's a good idea to ask at the start of every problem, "does an existing well-performing model exist for my problem?"
- When using a pretrained model, it's important that your custom data be formatted/preprocessed in the same way that the original model was trained on, otherwise you may get degraded performance.
- The same goes for predicting on custom data, ensure your custom data is in the same format as the data your model was trained on.
- There are several different places to find pretrained models from the PyTorch domain libraries, HuggingFace Hub and libraries such as timm (PyTorch Image Models).

| Method             | What happens                                          | Best when             |
| ------------------ | ----------------------------------------------------- | --------------------- |
| Feature extraction | Freeze most pretrained layers, train only final layer | Small custom dataset  |
| Fine-tuning        | Unfreeze some/all layers and train them too           | Larger custom dataset |


```python
# Feature Extraction

import torch
import torch.nn as nn
import torchvision.models as models

num_classes = 5

model = models.resnet18(weights="IMAGENET1K_V1")

# Freeze all pretrained layers
for param in model.parameters():
    param.requires_grad = False

# Replace final layer
model.fc = nn.Linear(model.fc.in_features, num_classes)

# Only classifier parameters are trained
optimizer = torch.optim.Adam(model.fc.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()

# This is safer when you have little data 
# because the model keeps the general ImageNet knowledge.
```

```python
# Fine-tuning
model = models.resnet18(weights="IMAGENET1K_V1")

# Replace final layer
model.fc = nn.Linear(model.fc.in_features, num_classes)

# Option A: fine-tune the whole model
for param in model.parameters():
    param.requires_grad = True

optimizer = torch.optim.Adam(model.parameters(), lr=1e-5)
criterion = nn.CrossEntropyLoss()
```

Small dataset  → feature extraction
Medium dataset → unfreeze final block
Large dataset  → fine-tune more layers or whole model
