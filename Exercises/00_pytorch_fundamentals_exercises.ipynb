### 4. Set the random seed to `0` and do 2 & 3 over again.

The output should be:
```
(tensor([[1.8542],
         [1.9611],
         [2.2884],
         [3.0481],
         [1.7067],
         [2.5290],
         [1.7989]]), torch.Size([7, 1]))
```

````python
# Set manual seed
RANDOM_SEED = 0
# Create two random tensors
torch.manual_seed(seed=RANDOM_SEED)
random_tensor_A = torch.rand(7, 7)
torch.manual_seed(seed=RANDOM_SEED)
random_tensor_B = torch.rand(1, 7)

# Matrix multiply tensors
result = torch.matmul(random_tensor_A, random_tensor_B.T)
result
````

### 5. Speaking of random seeds, we saw how to set it with `torch.manual_seed()` but is there a GPU equivalent? (hint: you'll need to look into the documentation for `torch.cuda` for this one)

````python
# Set random seed on the GPU
cuda = torch.device("cuda:0")
if torch.cuda.is_available():
  print(True)
else: 
  print('Not available')
  
torch.cuda.manual_seed(1234)
random_tensor_A = torch.rand(7, 7, device=cuda)
torch.cuda.manual_seed(1234)
random_tensor_B = torch.rand(1, 7, device=cuda)
random_tensor_A == random_tensor_B
````


### 10. Make a random tensor with shape `(1, 1, 1, 10)` and then create a new tensor with all the `1` dimensions removed to be left with a tensor of shape `(10)`. Set the seed to `7` when you create it and print out the first tensor and it's shape as well as the second tensor and it's shape.

The output should look like:

```
tensor([[[[0.5349, 0.1988, 0.6592, 0.6569, 0.2328, 0.4251, 0.2071, 0.6297,
           0.3653, 0.8513]]]]) torch.Size([1, 1, 1, 10])

        0.8513]) torch.Size([10])
```

````python
# Set seed
torch.manual_seed(42)

# Create random tensor
random_tensor = torch.rand(1, 1, 1, 10)

# Remove single dimensions
random_tensor_2 = random_tensor.squeeze()
torch.manual_seed(7)
# Print out tensors and their shapes
print(random_tensor, random_tensor.shape)
print(random_tensor_2, random_tensor_2.shape)
````
